## Problem Statement
Agricultural performance varies with seasonal environmental conditions, farming practices, resource availability and market conditions. This project analyzes the supplied agricultural dataset to identify meaningful seasonal patterns, trends, relationships and variations.

## Objectives
- Explore and understand the dataset
- Clean and prepare the data
- Compare agricultural performance across seasons
- Investigate environmental, resource and economic relationships
- Apply statistical and visualization techniques
- Develop evidence-based conclusions and recommendations

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
df.head()
```

```
print("Shape:", df.shape)
print(df.info())
print(df.describe(include="all").T)
```

```
missing = df.isna().sum()
print(missing[missing > 0])

numeric_cols = df.select_dtypes(include=np.number).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:", df.isna().sum().sum())
```

```
season_summary = df.groupby("Season").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield=("Yield_Tonnes_Ha","mean"),
    Avg_Profit=("Profit_INR","mean"),
    Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3","mean"),
    Avg_Disease_Risk=("Disease_Pest_Risk_pct","mean")
).round(2)
season_summary
```

```
season_summary["Avg_Yield"].plot(kind="bar", title="Average Yield by Season")
plt.ylabel("Tonnes/Ha")
plt.tight_layout()
plt.show()
```

```
irrigation_summary = df.groupby("Irrigation_Method").agg(
    Avg_Yield=("Yield_Tonnes_Ha","mean"),
    Avg_Profit=("Profit_INR","mean"),
    Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3","mean")
).round(2)
irrigation_summary
```

```
crop_yield = df.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)
crop_yield.plot(kind="bar", title="Average Yield by Crop")
plt.ylabel("Tonnes/Ha")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```
corr_cols = [
    "Rainfall_mm","Avg_Temperature_C","Humidity_pct","Soil_Moisture_pct",
    "Fertilizer_kg_ha","Pesticide_Litre_ha","Seed_Quality_Score",
    "Water_Efficiency_t_per_1000m3","Yield_Tonnes_Ha","Profit_INR"
]
corr = df[corr_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

## Key Findings
- **Kharif** has the highest average yield at **5.64 tonnes/ha**.
- Kharif has the highest average profit at approximately **₹178,915 per farm record**.
- **Zaid** has the lowest average profit at approximately **₹-24,805**.
- **Drip irrigation** has the highest average yield among the four irrigation methods at **6.62 tonnes/ha**.
- **Rainfed** has the highest water-efficiency measure at **7.56 tonnes per 1,000 m³**.
- The strongest positive correlation with yield is **Water_Efficiency_t_per_1000m3 (r ≈ 0.915)**.
- The highest-yield crop is **Sugarcane**, averaging **46.94 tonnes/ha**.
